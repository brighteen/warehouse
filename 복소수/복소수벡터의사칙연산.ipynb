```python
# 셀 1: 라이브러리 로드 및 공통 시각화 함수 정의
import matplotlib.pyplot as plt
import numpy as np

def init_complex_plane(xlim=(-6, 6), ylim=(-6, 6)):
    """복소평면의 기본 축과 격자를 설정하는 함수"""
    plt.figure(figsize=(7, 7))
    plt.axhline(0, color='black', linewidth=1.2)  # 실수축
    plt.axvline(0, color='black', linewidth=1.2)  # 허수축
    plt.grid(True, which='both', linestyle='--', alpha=0.5)
    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.xlabel('Real Axis (실수축)', fontsize=11)
    plt.ylabel('Imaginary Axis (허수축)', fontsize=11)

def draw_vector(z, color, label, start=[0, 0], linestyle='-'):
    """복소수를 평면 위에 화살표 벡터로 그리는 함수"""
    plt.quiver(start[0], start[1], z.real, z.imag, 
               angles='xy', scale_units='xy', scale=1, 
               color=color, label=label, linestyle=linestyle, alpha=0.8)
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] ='Malgun Gothic'
plt.rcParams['axes.unicode_minus'] =False
```

```python
# 셀 2: 복소수의 덧셈 시각화
z1 = 2 + 3j
z2 = 3 + 1j
z_add = z1 + z2

init_complex_plane()

# 1. 원점에서 출발하는 두 복소수 벡터
draw_vector(z1, 'blue', f'z1 = {z1}')
draw_vector(z2, 'green', f'z2 = {z2}')

# 2. 평행사변형 법칙을 보여주기 위해 z1의 끝점에서 z2만큼 이동하는 점선 벡터
draw_vector(z2, 'green', None, start=[z1.real, z1.imag], linestyle='--')

# 3. 덧셈 결과 벡터
draw_vector(z_add, 'red', f'z1 + z2 = {z_add}')

plt.title('Complex Addition (복소수 덧셈: 평행사변형 구조)', fontsize=13, pad=15)
plt.legend(loc='upper left')
plt.show()
```

```python
# 셀 3: 복소수의 뺄셈 시각화
z1 = 5 + 4j
z2 = 2 + 3j
z_sub = z1 - z2

init_complex_plane()

# 원점 출발 벡터
draw_vector(z1, 'blue', f'z1 = {z1}')
draw_vector(z2, 'green', f'z2 = {z2}')

# z2에서 z1으로 향하는 변위 벡터 (뺄셈의 본질)
draw_vector(z_sub, 'purple', None, start=[z2.real, z2.imag], linestyle='--')

# 뺄셈 결과 벡터 (원점 정렬)
draw_vector(z_sub, 'red', f'z1 - z2 = {z_sub}')

plt.title('Complex Subtraction (복소수 뺄셈)', fontsize=13, pad=15)
plt.legend(loc='upper left')
plt.show()
```

```python
# 셀 4: 복소수의 곱셈 시각화 (회전과 확대)
# z1: 크기 약 1.414 (sqrt(2)), 각도 45도
z1 = 1 + 1j 
# z2: 크기 2.0, 각도 90도
z2 = 0 + 2j 
# z_mul: 크기 약 2.828 (2*sqrt(2)), 각도 135도 (45 + 90)
z_mul = z1 * z2

init_complex_plane(xlim=(-4, 4), ylim=(-4, 4))

draw_vector(z1, 'blue', f'z1 = {z1} (45°)')
draw_vector(z2, 'green', f'z2 = {z2} (90°)')
draw_vector(z_mul, 'red', f'z1 * z2 = {z_mul} (135°)')

# 원점 중심의 회전 궤적 표현 (호 그리기)
theta = np.linspace(np.pi/4, np.pi/4 + np.pi/2, 100)
r = np.abs(z1)
plt.plot(r * np.cos(theta), r * np.sin(theta), 'k--', alpha=0.5, label='Rotation Path')

plt.title('Complex Multiplication (복소수 곱셈: 각도의 합과 크기의 곱)', fontsize=13, pad=15)
plt.legend(loc='upper right')
plt.show()
```

```python
# 셀 5: 복소수의 나눗셈 시각화 (역회전과 축소)
# z1: 크기 약 2.828, 각도 135도
z1 = -2 + 2j
# z2: 크기 2.0, 각도 90도
z2 = 0 + 2j
# z_div: 크기 약 1.414, 각도 45도 (135 - 90)
z_div = z1 / z2

init_complex_plane(xlim=(-4, 4), ylim=(-4, 4))

draw_vector(z1, 'blue', f'z1 = {z1} (135°)')
draw_vector(z2, 'green', f'z2 = {z2} (90°)')
draw_vector(z_div, 'red', f'z1 / z2 = {z_div} (45°)')

plt.title('Complex Division (복소수 나눗셈: 각도의 차와 크기의 나눗셈)', fontsize=13, pad=15)
plt.legend(loc='upper right')
plt.show()
```
