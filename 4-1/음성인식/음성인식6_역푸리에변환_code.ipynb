```python
import numpy as np

def DFT(x):

    N = len(x)
    n = np.arange(N)
    k = n.reshape((N,1))
    e = np.exp(-2j * np.pi * k * n / N)
    X = np.dot(e, x)
    return X

def IDFT(X):

    N = len(X)
    k = np.arange(N)
    n = k.reshape((N,1))
    e = np.exp(+2j * np.pi * k * n / N)
    x = (1/N) * np.dot(e, X)
    return x

# (1) 임의 신호 생성
x_original = np.array([0, 1, 2, 3])

# (2) DFT 계산
X_dft = DFT(x_original)
print("DFT 계수 X_dft:", X_dft)

# (3) IDFT로 복원
x_recovered = IDFT(X_dft)
print("IDFT로 복원한 신호 x_recovered:", x_recovered)

# (4) 원 신호와 비교
print("원 신호와의 차이 (오차):", np.round(x_recovered - x_original, 3))
```
