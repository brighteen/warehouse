```python
result = model.transcribe(audio_path)
print(result["text"])

tokenizer = whisper.tokenizer.get_tokenizer(multilingual=True)
print('tokenizer:', tokenizer)
```

```python
for i, segment in enumerate(result["segments"]):
    print(f"\n[Segment {i}] {segment['start']:.2f}s ~ {segment['end']:.2f}s")
    print(f"  텍스트: {segment['text']}")
    print(f"  토큰ID: {segment['tokens']}")
    print(f"  시작ID: {tokenizer.timestamp_begin}")

    for token_id in segment["tokens"]:
        text = tokenizer.decode([token_id])
        if token_id >= tokenizer.timestamp_begin:
            ts = (token_id - tokenizer.timestamp_begin) * 0.02
            print(f"    ID {token_id}: [timestamp {ts:.2f}s]")
        else:
            print(f"    ID {token_id}: '{text}'")
```
